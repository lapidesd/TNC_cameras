```python
import os

dirs = os.listdir('../Data/Eddy_covariance')
dirs = [d for d in dirs if (not (d.endswith('Store'))) & (not (d.endswith('zip')))]
dirs
```

```python
fuse = {d:0 for d in dirs}
for d in dirs[:]:
    print(d)
    flist = os.listdir('../Data/Eddy_covariance/'+d)
    for f in flist:
        try:
            dd = pd.read_csv('../Data/Eddy_covariance/'+d+'/'+f,skiprows=2)
            if 'TIMESTAMP_START' not in dd.columns:
#                 print('oops')
                dd = pd.read_csv('../Data/Eddy_covariance/'+d+'/'+f)#,skiprows=2)
                if 'TIMESTAMP_START' not in dd.columns:
#                     print('what?')
                    continue
        except:
            continue
        if dd.TIMESTAMP_START.values[0]>1000000000:
#             print(f,len(dd))
            LElist = [c for c in dd.columns if 'LE' in c]
            if len(LElist)>0:
                dd = dd[dd[LElist[0]].astype('int')>-109]
#                 print(dd[['TIMESTAMP_START']+LElist].head())
                fuse[d] = f
                print('   '+f)
```

```python
LEdata = []
for d in dirs[:]:
    f = fuse[d]
    try:
        dd = pd.read_csv('../Data/Eddy_covariance/'+d+'/'+f,)#skiprows=1)
        dd['datetime'] = dd.TIMESTAMP_START.astype('str')
    except:
        dd = pd.read_csv('../Data/Eddy_covariance/'+d+'/'+f,skiprows=2)
        dd['datetime'] = dd.TIMESTAMP_START.astype('str')
    dd['datetime'] = dd.datetime.apply(lambda s:s[:4]+'-'+s[4:6]+'-'+s[6:8]+' '+s[8:10]+':'+str(int(s[10:])+15))
    dd['datetime'] = pd.to_datetime(dd.datetime)
    dd.set_index('datetime',inplace=True)
    LElist = [c for c in dd.columns if 'LE' in c]
    LElens = {L:0 for L in LElist}
    for LEcol in LElist:
        LElen = len(dd[dd[LEcol].astype('int')>-109])
        LElens[LEcol]=LElen
    LEcol_use = max(LElens, key=LElens.get)
    dd = dd[[LEcol_use]].rename(columns={LEcol_use:(d.split('-')[1])})
    dd = dd[dd[d.split('-')[1]].astype('float')>-1000]*0.0353 
    LEdata.append(dd)
LEdata = pd.concat(LEdata,)
```
